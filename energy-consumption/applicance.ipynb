```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
appliances_energy_prediction = fetch_ucirepo(id=374)

# data (as pandas dataframes)
X = appliances_energy_prediction.data.features
y = appliances_energy_prediction.data.targets

# metadata
# print(appliances_energy_prediction.metadata)

# variable information
print(appliances_energy_prediction.variables)
```

```python
import pandas as pd

df = pd.concat([X, y], axis=1)
df.head()
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["date"].head(10)
```

```python
df["date"] = df["date"].str.replace(
    r"(\d{4}-\d{2}-\d{2})(\d{2}:\d{2}:\d{2})",
    r"\1 \2",
    regex=True
)
```

## Explore the target variable

```python
df["Appliances"].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df["Appliances"], bins=50)
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.plot(df["date"], df["Appliances"])
plt.xlabel("Date")
plt.ylabel("Energy Consumption (Wh)")
plt.show()
```

## Plot the time series

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(14,10))
sns.heatmap(corr, cmap="coolwarm")
plt.show()
```

```python
corr["Appliances"].sort_values(ascending=False)
```

```python
df["hour"] = df["date"].dt.hour
df["day"] = df["date"].dt.day
df["month"] = df["date"].dt.month
df["dayofweek"] = df["date"].dt.dayofweek
df["weekend"] = (df["dayofweek"] >= 5).astype(int)
```

```python
df["lag_1"] = df["Appliances"].shift(1)
df["lag_2"] = df["Appliances"].shift(2)
df["lag_6"] = df["Appliances"].shift(6)
df["lag_12"] = df["Appliances"].shift(12)
df["lag_144"] = df["Appliances"].shift(144)  # 24 hours earlier
```

```python
df = df.dropna()
```

```python
X = df.drop(columns=["Appliances", "date"])
y = df["Appliances"]
```

```python
split = int(len(df) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

## Feature scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Train the model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

print(f"MAE : {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²  : {r2:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,5))

plt.plot(y_test.values[:300], label="Actual")
plt.plot(y_pred[:300], label="Predicted")

plt.legend()
plt.xlabel("Time")
plt.ylabel("Energy Consumption")
plt.title("Linear Regression Predictions")
plt.show()
```

```python
coef_df = (
    pd.DataFrame({
        "Feature": X.columns,
        "Coefficient": model.coef_
    })
    .sort_values("Coefficient", key=abs, ascending=False)
)

coef_df.head(10)
```

## Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
y_pred_rf = rf.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred_rf)
rmse = mean_squared_error(y_test, y_pred_rf) ** 0.5
r2 = r2_score(y_test, y_pred_rf)

print(f"MAE : {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²  : {r2:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,5))

plt.plot(y_test.values[:300], label="Actual")
plt.plot(y_pred_rf[:300], label="Random Forest")

plt.legend()
plt.xlabel("Time")
plt.ylabel("Energy Consumption (Wh)")
plt.title("Random Forest Predictions")
plt.show()
```
